```python
import whisper
import os
import json
from datetime import datetime

# Load the Whisper model
print("Loading Whisper model...")
model = whisper.load_model("base")
print("✓ Model loaded successfully!\n")

# Create jsons folder if it doesn't exist
os.makedirs('jsons', exist_ok=True)

# List only audio files from audio_remaining folder
audios = [f for f in os.listdir('/content/audio_remaining')
          if os.path.isfile(f'/content/audio_remaining/{f}')
          and f.endswith(('.mp3', '.wav', '.m4a', '.flac', '.ogg'))]

print(f"Found {len(audios)} audio files in /content/audio_remaining\n")
print("="*60)

start_time = datetime.now()

# Process each audio file
for idx, audio in enumerate(audios, 1):
    if("_" in audio):
        # Extract number and title from filename
        number = audio.split("_")[0]
        title = audio.split("_")[1]

        file_start = datetime.now()
        print(f"\n[{idx}/{len(audios)}] Processing: {audio}")
        print(f"Number: {number}, Title: {title.rsplit('.', 1)[0]}")
        print(f"⏳ Starting transcription at {file_start.strftime('%H:%M:%S')}...")

        try:
            # Transcribe audio
            result = model.transcribe(
                audio=f"/content/audio_remaining/{audio}",
                language="hi",
                task="translate",
                word_timestamps=False,
                verbose=False  # Less verbose output
            )

            file_end = datetime.now()
            duration = (file_end - file_start).total_seconds()
            print(f"✓ Transcription completed in {duration:.1f} seconds")

            # Create chunks with metadata
            chunks = []
            for segment in result["segments"]:
                chunks.append({
                    "number": number,
                    "title": title.rsplit('.', 1)[0],
                    "start": segment["start"],
                    "end": segment["end"],
                    "text": segment["text"]
                })

            # Prepare final data structure
            chunks_with_metadata = {
                "chunks": chunks,
                "text": result["text"]
            }

            # Create filename without extension
            audio_name = audio.rsplit('.', 1)[0]
            filepath = f"jsons/{audio_name}.json"

            # Save to JSON file
            with open(filepath, "w", encoding='utf-8') as f:
                json.dump(chunks_with_metadata, f, indent=2, ensure_ascii=False)

            print(f"✓ Saved: {filepath}")
            print(f"  - Chunks: {len(chunks)}")
            print(f"  - Size: {os.path.getsize(filepath):,} bytes")
            print(f"  - Total time: {duration:.1f}s")

        except Exception as e:
            print(f"✗ ERROR processing {audio}: {str(e)}")
            continue
    else:
        print(f"\n[{idx}/{len(audios)}] Skipping {audio} (no underscore in filename)")

# Summary
end_time = datetime.now()
total_duration = (end_time - start_time).total_seconds()

print("\n" + "="*60)
print("PROCESSING COMPLETE!")
print(f"Total time: {total_duration/60:.1f} minutes")
print("="*60)

# List all created JSON files
json_files = sorted(os.listdir('jsons'))
print(f"\nTotal JSON files created: {len(json_files)}\n")

for f in json_files:
    size = os.path.getsize(f'jsons/{f}')
    print(f"✓ {f} ({size:,} bytes)")

print("\n" + "="*60)
print("All files saved in: /content/jsons/")
print("="*60)
```

```python
import whisper
import os
import json

# Create jsons folder
os.makedirs('jsons', exist_ok=True)

# List only audio files (not folders)
audios = [f for f in os.listdir('/content/audios')
          if os.path.isfile(f'/content/audios/{f}')
          and f.endswith(('.mp3', '.wav', '.m4a', '.flac'))]

print(f"Found {len(audios)} audio files")

for audio in audios:
    if("_" in audio):
        number = audio.split("_")[0]
        title = audio.split("_")[1]

        print(f"\nProcessing: {audio}")
        print(f"Number: {number}, Title: {title.rsplit('.', 1)[0]}")

        result = model.transcribe(audio=f"/content/audios/{audio}",
                                language="hi",
                                task="translate",
                                word_timestamps=False)

        chunks = []
        for segment in result["segments"]:
            chunks.append({
                "number": number,
                "title": title.rsplit('.', 1)[0],
                "start": segment["start"],
                "end": segment["end"],
                "text": segment["text"]
            })

        chunks_with_metadata = {"chunks": chunks, "text": result["text"]}

        audio_name = audio.rsplit('.', 1)[0]
        filepath = f"jsons/{audio_name}.json"

        print(f"Saving to: {filepath}")
        print(f"Data to save: {len(chunks)} chunks")

        with open(filepath, "w", encoding='utf-8') as f:
            json.dump(chunks_with_metadata, f, indent=2, ensure_ascii=False)

        print(f"✓ File saved successfully!")

# List all created JSON files
print("\n" + "="*50)
print("SUMMARY - JSON files created:")
print("="*50)
json_files = os.listdir('jsons')
for f in json_files:
    size = os.path.getsize(f'jsons/{f}')
    print(f"✓ {f} ({size:,} bytes)")
```
